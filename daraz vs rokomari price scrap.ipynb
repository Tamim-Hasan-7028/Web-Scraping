```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
import pandas as pd
```

```python
driver = webdriver.Chrome()
driver.maximize_window()

rokomari_price = []
daraz_price = []
#entering website
def rokomari():
    url = 'https://www.rokomari.com/'
    driver.get(url)
    
    time.sleep(3)
    
    search_bar = driver.find_element(By.XPATH, '//*[@id="js--search"]')
    search_bar.click()
    
    search_query = "QCY HT05"
    search_bar.send_keys(search_query)
    time.sleep(2)
    
    # Press Enter to search
    search_bar.send_keys(Keys.ENTER)
    time.sleep(3)
    
    driver.find_element(By.XPATH,'//*[@id="js--browse-sidebar-form"]/div[2]/div[2]/div[3]').click()
    time.sleep(2)
    
    # driver.find_element(By.XPATH,'//*[@id="js--specifications"]/div[2]/label').click()
    # time.sleep(3)
    
    # Scroll the element into view before clicking
    # checkbox = driver.find_element(By.ID, 'rating-4')
    # driver.execute_script("arguments[0].scrollIntoView();", checkbox)
    # checkbox.click()

    time.sleep(2)
    
    # driver.find_element(By.XPATH,'//*[@id="ts--desktop-filter-&-sort-container"]/div[6]/div[9]/div[2]/label').click()
    # time.sleep(3)
    
    
    # Target the parent element first
    element = driver.find_element(By.XPATH, "/html/body/div[7]/div/div/div/section[2]/div[2]/div/div[2]/div/a/div[2]/p[3]")
    
    # Extract the text from that element (this will include all text inside it, including the second text node)
    text_content = element.text

    # If needed, split the text to get the specific part you want
    # Example: If you want the second text node only
    text_parts = text_content.split("TK.")  # or split by spaces or commas based on the format
    # print(text_parts)
    desired_text = text_parts[2].strip()  # Get the second part of the text
    # print(desired_text)
    # Adjust as needed
    rokomari_price.append(desired_text)
    
    # print(rokomari_price)

def daraz():
    url = 'https://www.daraz.com.bd/'
    driver.get(url)
    
    time.sleep(5)
    
    search_bar = driver.find_element(By.XPATH, '//*[@id="q"]')
    search_bar.click()
    
    search_query = "QCY HT05"
    search_bar.send_keys(search_query)
    time.sleep(2)
    
    # Press Enter to search
    search_bar.send_keys(Keys.ENTER)
    time.sleep(3)
    
    driver.find_element(By.XPATH,'//*[@id="root"]/div/div[2]/div[1]/div/div[2]/div/div[1]/div[2]/div/div/div/a[1]').click()
    time.sleep(3)
    
    driver.find_element(By.XPATH,'//*[@id="root"]/div/div[2]/div[1]/div/div[2]/div/div[6]/div[2]/div[2]/span').click()
    time.sleep(3)
    
    driver.find_element(By.XPATH,'//*[@id="root"]/div/div[2]/div[1]/div/div[2]/div/div[7]/div[2]/div/div/label[1]/span[2]').click()
    time.sleep(3)
    
    # driver.find_element(By.XPATH,'//*[@id="root"]/div/div[2]/div[1]/div/div[1]/div[1]/div/div[2]/div/div[2]/div/div/span[2]/div').click()
    # time.sleep(3)
    
    # driver.find_element(By.XPATH,'//*[@id="root"]/div/div[2]/div[1]/div/div[1]/div[1]/div/div[2]/div/div[2]/div/div/span[2]/div').click()
    # time.sleep(3)
    
    
    price = driver.find_element(By.XPATH, '//*[@id="root"]/div/div[2]/div[1]/div/div[1]/div[3]/div[1]/div/div/div[2]/div[3]/span').text  # Adjust as needed
    daraz_price.append(price)
    # print(daraz_price)

rokomari()
daraz()

driver.quit()

df=pd.DataFrame({'Rokomari':rokomari_price,'Daraz':daraz_price})
print(df)
```
